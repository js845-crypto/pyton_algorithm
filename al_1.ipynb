```python
# 1번문제
num = int(input("학생번호 : "))
# ((num) -1
#1~10 -> 0
# 11~20 -> 10
leader_num = ((num-1)//10*10)+1

print(leader_num)
```

```python
# 2번문제

try:
    num = int(input("숫자를 입력하세요"))

    if num %2 ==0:
        print("짝수")
    else:
        print("홀수")
except ValueError as e:
    print("정수만 입력하세요")    
    # print(e) 실제 ValueError에 있는 error message, 사용자에게 적합한 안내는 아니라서 주석처리 함
```

```python
# 3번 문제
try:
    number = int(input("숫자를 입력하세요"))
    
    if number<0 :
        print(f"입력하신 숫자 {number}는 양수입니다")
    
    elif number ==0:
        print(f"입력하신 숫자{number}는 0 입니다")
    
    else:
        print(f"입력하신 숫자{number}는 양수 입니다")

except ValueError as e:
    print("정수만 입력하세요")

```

```python
try:
    person1 = int(input("나이를 입력하세요"))
    person2 = int(input("나이를 입력하세요"))
    
    if person1 and person2<20:
        print("20~29세만 출입가능합니다. 출입이 불가합니다")
        if person1 and person2>29:
            print("20~29세만 출입가능합니다. 출입이 불가합니다") 
        
    
    print("출입이 가능합니다")
except ValueError as e:
    print("나이는 정수만 입력하세요")
```

```python
# 13번 문제

for i in range(1,10):
    for dan in range(2,10):
        print(f"{dan} x {i} = {dan*i:>2}", end= "|") # {dan*i:>2} 우측정렬 의미
         # {dan*i:>2}이거 뒤에 end ="" 하면 옆으로 출력되고 없으면 기본 default라 줄바꿈임 end ="\n" 이므로
    print("", end ="\n") # print에선 end ="\n"가기본인데 굳이 넣었음


```
