```python
from collections import deque
dq = deque()
print(dq)
```

```python
# 오른쪽 data 추가는 append ,= stack
dq = deque()
dq.append(10)
dq.append(20)
dq.append(30)

print(dq)
```

```python
# 왼쪽 data 추가는 appendleft
dq = deque()
dq.append(20)
dq.append(30)
dq.appendleft(10)

print(dq)
```

```python
# 오른쪽 data 삭제는 pop , = stack
dq = deque([10, 20, 30])
x = dq.pop()

print(x)
print(dq)
```

```python
# 왼쪽 data 삭제는 popleft

dq = deque([10, 20, 30])
x = dq.popleft()

print(x)
print(dq)
```

```python
# 양방향 추가/ 삭제
dq = deque()

dq.append(10)
dq.append(20)
dq.appendleft(5)
dq.appendleft(1)

print(dq)

print(dq.pop())
print(dq.popleft())

print(dq)
```

```python
# rotate
dq = deque([1, 2, 3, 4, 5])
dq.rotate(1) # 우측으로 한칸 이동(회전)

print(dq) # deque([5, 1, 2, 3, 4])

# 위의 dq.rotate method와 같은 동작하는 코드, 회전 횟수가 두번이면 pop, pop , appendleft, appendleft 해야해서 여러번 회전일 땐 위의 dq.rotate method 사용하는게 편함
dq1 = deque([1,2,3,4,5])
x = dq1. pop()
dq1.appendleft(x)
print(dq1)


```

```python
dq = deque([1, 2, 3, 4, 5])
dq.rotate(-2) # 왼쪽으로 두 칸 회전

print(dq)

# stack이 필요할 때는 효율성 때문에 deque를 사용하진 않음.
# 그러나 queue가 필요할 땐 효율성 때문에 deque를 사용함.
# deque가 필요할 땐 deque 사용
# 자료구조 회전이 필요할 땐 시계방향 반시계 방향 둘다 필요하면 deque이고 시계방향일 땐 queue 사용. 그러나 효율성 때문에 시계 방향만 필요해도 deque를 사용할 수 있음
```

```python
# 최근 방분 기록 n개 저장
history = deque(maxlen=5) # maxlen=n n개만 저장

history.append("google")
history.append("naver")
history.append("youtube")
history.append("github")
history.append("chatgpt")

print(history)

history.append("python")

print(history)
# method를 사용하면 추가 및 삭제를 최대 갯수에 맞춰서 알아서 구현하고 칸 이동 정렬도 된다.
# 기본 모듈 안에 deque를 사용하는 이유는 deque이라는 자료구조 때문
# deque 편리성 method를 많이 제공(큐, 스택, 회전, 최대갯수,....)
```

```python
# 카드 게임(원카드, 우노, ..) 시뮬레이션

cards = deque([1, 2, 3, 4, 5])

while len(cards) > 1:
    print("버림 :", cards.popleft())
    cards.rotate(-1)
    # cards.append(cards.popleft()) 이렇게 해도 되지만 rotate method 사용하면 편함
    print("남은 카드 :", cards)

print("마지막 카드 :", cards[0])
```
