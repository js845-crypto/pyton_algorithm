```python
# 데이터 꺼내기 Deque
queue = ["A", "B", "C"]

print(queue.pop(0)) # 숫자를 안주면 stack에서 그냥 pop 즉 마지막 것이 뽑혔고 pop[0]하면 queue로서 첫번째가 나가짐. 파이썬에서 따로 구현할 필요 없음 list 메소드로 구현되있음
print(queue) # 첫번째 꺼를 빼면서 나머지 원소들을 한칸씩 밀어야해서 n이 증가하면 시간복잡도가 엄청나게 증가함. 단점이다.

print(queue.pop(0))
print(queue)

print(queue.pop(0))
print(queue)
```

```python
queue = ["A", "B", "C"]
print(queue[0])
print(queue[1:])

print(queue[1])
print(queue[2:])

print(queue[2])
print(queue[3:]) # 이렇게 하면 실제 queue처럼 작동시킬 수 있으나 print할 때 인덱스를 계속 코드로 바꿔야함. 이거를 반복문으로 구현해보자
```

```python
queue = ["A", "B", "C"]
pop_index = 0
while pop_index < len(queue):
    print(queue[pop_index])
    print(queue[pop_index+1:])
    queue = queue[pop_index+1:] # slicing한 것을 queue에 다시 할당하면 위의 pop 메소드에서 한칸씩 데이터 이동하는것보다 시간복잡도는 좋아지나 메모리가 더 필요해진다, trade off 관계임
    pop_index += 1         

print(queue) # 이렇게 반복문으로 구현했지만 코드가 엄청 복잡해진다. list안에 덱이나 큐 등이 모듈로 사용하는게 있다.

```

```python
# stack Vs queue
stack = []

stack.append("A")
stack.append("B")
stack.append("C")

while len(stack) > 0:
    print(stack.pop(), end=" ")

print("\n")


queue = []

queue.append("A")
queue.append("B")
queue.append("C")

while len(queue) > 0:
    print(queue.pop(0), end=" ")
```

```python
# queue에서는 pop하고서 앞으로 데이터를 한칸 씩 데이터가 많을수록 비효율적이다.
#실제 알고리즘에서는 큐를 구현할 때 collections모듈의 deque를 사용한다.
# deque의 popleft() 메서드를 사용하면 맨 앞의 데이터를 이동 없이 빠르게 꺼낼 수 있어 알고리즘 문제에서는 deque를 사용하는 것이 표준적인 방법이다

from collections import deque

queue = deque()

queue.append("A")
queue.append("B")
queue.append("C")

print(queue.popleft())
print(queue) # queue라고 적었지만 실제론 deque 임을 잊지말자
print(queue.pop()) # 우측을 빼는 것은 그냥 pop() 즉 stack 사용하면 됨
print(queue) # 실제로 빠지지 않았지만 빠진 것처럼 보인다. deque를 사용해서 queue 구현을 실무에서 많이한다. list queue 메소드로 하는 건 시간복잡도가 너무 증가함.

```
