```python
import re

text = "a7bc123"
result = re.findall(r"^[0-9]", text) # ^ 문자가 괄호 밖에 있으면 0~9숫자중에 하나로 시작되는지를 체크하는 것, 괄호 안에 있어야 그걸 제외한 문자들을 탐색 반환
print(result)
```

```python
import re

text = "cat ct cot cut coot"
result = re.findall(r"c.t", text) # "c.t" 패턴의 문자(c로 시작, t로 끝나는 문자열길이) , .는 아무 문자를 의미
print(result)
```

```python
import re

# findall함수에서는 ^패턴을 사용해도 의미있는 반환값을 얻을 수 없다(^패턴은 전체 문자열을 하나로 다루기 때문)
text = "Python Python python" # "Python Python python" 이 문자열의 시작은 P이다. 세개의 문자열이 아니라 하나의 문자열임을 유의
print(re.findall(r"^Py", text)) # ["Py"]

text = "python Python python" # "python Python python" 문자열은 p로 시작해서 아무것도 반환되지 않음
print(re.findall(r"^Py", text)) # [] 

```

```python
texts = ["Python Python python","python Python python" ]
for t in texts: # 두개의 문자열을 각각 순회한다 즉 두번 반복.
    print(re.findall(r"^Py", t))
```

```python
import re

text = "Python Python python"
print(re.search(r"^Py", text)) # <re.Match object; span=(0, 2), match='Py'> span은 Py문자의 위치 0에서 2-1에 위치
print(re.search(r"^py", text)) # None
print(re.search(r"py", text)) 
print(re.match(r"py", text)) # match는 ^이 없어도 자체가 이미 시작하는걸 찾아서 ^가 포함되있는 것
```

```python
import re

text = "Python"
# 끝나는 문자 찾으려면 search 사용하고 $ 문자를 뒤에 달아준다
print(re.search(r"hon$", text)) # <re.Match object; span=(3, 6), match='hon'>
print(re.search(r"abc$", text)) # None
```

```python
# * 앞에 문자는 0번이상 여러번 반복될 수 있다. 최소 개수 0개
# + 앞에 문자는 1번이상 여러번 반복될 수 있다. 최소 개수 1개

import re

text = "Pyyyyython"

# 시작이 Py 찾기
print(re.search(r"^Py", text)) # <re.Match object; span=(0, 2), match='Py'>
# 시작이 Pyyyyy.... 찾기
print(re.search(r"Py*", text)) # <re.Match object; span=(0, 6), match='Pyyyyy'> *앞에 문자 y가 0번이상 이므로 y가 없어도 탐색해서 반환!
print(re.search(r"Py+", text)) # +앞에 문자가 1번이상 나와야함



```

```python
# fullmatch()함수 : 문자열 전체가 패턴과 일치해야 함 

import re

texts = ["3","123", "", "abc", "12a"]

for t in texts:
    print(t, "->", bool(re.fullmatch(r"\d+", t))) # \d+ : 숫자가 1개 이상 있어야 한다.
```

```python
import re

texts = ["3","123", "", "abc", "12a"]

for t in texts:
    print(t, "->", bool(re.fullmatch(r"\d*", t))) # \d* 숫자 0개이상, 즉 없어도 된다. 빈 문자열도 True
```

```python
# 수량자 ?
# 앞 문자가 0개 또는 1개 존재해야 매칭됨.
# "ab?c" : b가 없거나 1개 있거나 -> "abc"(O) 와 "ac"(O)는 매칭,  "abbc"(X) b가 2개라서 매칭 안됨

import re

words = [
    "ae",
    "abc",
    "abbcd",# 수량자 '?'는 2개 있어도 안된다 1개가 있거나 아예 없어야함 
    "abbbcde",
    "acde"
]

for word in words:
    print(re.match(r"ab?c", word))
# None
# <re.Match object; span=(0, 3), match='abc'>
# None
# None
# <re.Match object; span=(0, 2), match='ac'>
```

```python
import re

text = "Py@thon 123_xyz"

print(re.findall(r"\d", text)) # 숫자
print(re.findall(r"\D", text)) # 숫자가 아닌 문자
print(re.findall(r"\w", text))
print(re.findall(r"\W", text)) # 특수문자, 공백
print(re.findall(r"\s", text)) # 공백
print(re.findall(r"\S", text)) # 공백이 아닌 문자
```

```python
# 수량자 {n}
import re

text = "1,23,456,7890,77a8888"
print(re.findall(r"\d{3}", text)) # 숫자문자가 3번연속 나오는 구간만 반환 리스트에 추가
# ['456', '789', '888']
```

```python
# 수량자 {n}
import re

text = "1,23,456,7890,77a8888"
print(re.search(r"\d{3}", text)) # 숫자문자가 3번연속 일치하는지만 찾는거기 때문에 search는 처음 찾는순간 뒤에는 안찾는다 이것을 유의
# ['456', '789', '888']
```

```python
# 앞 문자가 최소 n번, 최대 m번 반복된다.
import re

text = "1 12 123 1234 12345"

print(re.findall(r"\d{2,4}", text)) # ['12', '123', '1234', '1234']
```

```python
# 여러 패턴 중 하나와 일치하면 된다.
import re

text = "cat, dog, bird, cats, dogs"

print(re.findall(r"cat|dog", text)) # ['cat', 'dog', 'cat', 'dog']

```

```python
import re

phones = [
    "02-1234-5678",
    "010-1234-5678",
    "010-123-5678",
    "010-1234-11111"
]

# "숫자3시작-숫자4-숫자4끝나는" 찾기
for p in phones:
    result = re.search(r"^\d{3}-\d{4}-\d{4}$", p)
    if result != None:
        print(result)
        print(result.group())
```
