```python
# search에다가 ^추가하면 match랑 같아진다. search 자체는 default가 처음부터가 아님. match는 시작하는 문자열을 찾는것임을 명심
```

```python
import re
text = "Python"
result = re.match(r"Py", text)
print(result)
print(result, result.group()) # group()함수 매칭된 결과물을 반환
```

```python
# 시작이 아닌 중간쪽을 탐색하고 싶으면 search 사용, 그러나 search는 문자열 여러개가 있는 list에서 "123ABC456" , "34AC12"이런식이면 앞의 123, 34만 찾는다 문자열의 뒤는 못찾음
# re.search는 문자열 전체를 검색하여 일치하는지 아닌지를 반환하기 때문에 다시 말하지만 처음에 찾으면 뒤에는 있어도 안찾고 반환은 첫번째만 함.
import re

text = "Python"
print(re.search(r"th", text)) # <re.Match object ...>
```

```python
# re.findall() 모든 검색 결과를 리스트로 반환한다.
import re

text = "apple banana orange"
result = re.findall(r"a", text)
print(result) # ['a', 'a', 'a', 'a', 'a']
```

```python
# re.sub(패턴, 변경문자, 문자열)
# 숫자 문자 제거
import re

text = "A10 B20 C30"
result = re.sub(r"\d", "", text) # text안에서 숫자들을 전부 공백으로 치환
print(result)
```

```python
import re

text = "A10 B20 C30"
result = re.sub(r"\D", "0", text) # text안에서 숫자가 아닌 것들을 전부 0으로 치환, 이러면 공백도 0으로 치환됨
# "\D" 이거랑 [^\d]랑 같다. 공백은 냅두려면 이걸 이용하자
result = re.sub(r"[^\d\s]", "0", text) # 이러면 공백이 아니고 숫자가 아닌 애를 0으로 치환
print(result) 
```

```python
# 상품명에서 가격만 출력. 숫자만 뽑아내자
import re

products = [
    "사과 3500원",
    "배 4200원",
    "수박 18000원"
]

for product in products:
    price = re.search(r"\d+", product)
    print(price.group())
```

```python
import re

products = [
    "사과 3500원",
    "배 4200원",
    "수박 18000원"
]
price_list = []

for product in products:
    name = re.search(r"[가-힣]+", product)
    price = re.search(r"\d+", product)
    # print(price.group())
    price_list.append({name.group() : int(price.group())})

print(price_list)
# [{"사과" : 3500}, {"배" : 4200}, ...]

```

```python
import re

emails = [
    "python@네이버.com",
    "kim123@gmail.com",
    "abc@company.co.kr", # [A-Za-z0-9.-]+ 이거 때문에 company.co 가 허용된다
    "hello",
    "@gmail.com"
]

pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$" # [.] 이렇게 괄호 안의 점은 그냥 . 문자를 의미한다. 괄호 밖 .이 모든 문자를 의미하는 것 헷갈리지 말자

for email in emails:
    if re.fullmatch(pattern, email):
        print(email, "정상")
    else:
        print(email, "오류")
```

```python
import re

urls = [
    "https://www.google.com",
    "http://naver.com",
    "https://python.org/docs",
    "google.com"
]

pattern = r"^https?://([\w-]+\.)+[\w-]+(/[^\s]*)?$"
# ^https? : "http" or "https"로 시작하는걸 찾는 것
# [\w-]+ : [\w-]+는 알파벳(\w) or -가 여러개 온다는 의미, 괄호로 묶인 것들이 여러개 반복되야한다 ex) https://abc, https://ab-c, https://-a-
# ([\w-]+\.)+ : https://abc.abc.abc. https://www.naver.www.www. 이런식으로 된다는 것. ()는 group화 하는 것임을 명심. 
# (/[^\s]*)? : /와 공백이 아닌 문자가 0번이상(*) 있어도 그만 없어도그만 (문자 ?)

for url in urls:
    if re.fullmatch(pattern, url):
        print(url, "정상")
    else:
        print(url, "오류")
```

```python
# 앞자리는 6자리 

# 문자 

# 주민번호 뒷자리 1~4문자 하나 : 외국인 제외, 돌아가신분들 제외 

# 뒷자는 6자리
import re

nums = [
    "901010-1234567",
    "000101-4234567",
    "9010101234567",
    "90101-1234567"
]

pattern = r"^\d{6}-[1-4]\d{6}$"

for n in nums:
    if re.fullmatch(pattern, n):
        print(n, "형식 정상")
    else:
        print(n, "오류")
```
