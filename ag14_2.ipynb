```python
print("Hello\nworld\\") # \\ 는 escape 문자 \를 사용 안하고 문자 그대로 사용

# print("\d") 는 error가 발생함. \d라는 escape 문자가 없음, \\를 사용하거나 print(r"")을 사용해야함
print("\\d")
print(r"\d")
```

```python
#  []안에 있는 문자 중 하나와 일치한다.

# re모듈의 findall(정규표현식, text) 함수는 text문자열의 0부터 끝까지 정규표현식이 일치하는 문자열들을 반환한다. 

import re

text = "apple banana cat"
result = re.findall(r"[abc]", text) # "apple banana cat"를 0부터 끝까지 "a", "b", "c" 중 일치하면 반환되는 리스트에 추가
print(result)
# ['a', 'b', 'a', 'a', 'a', 'c', 'a'] 가 출력됨 , 없으면 result에 반환되는게 없으므로 print로 출력 안됨

result2 = re.findall(r"apple", text) # 괄호를 빼고 "문자열" 하면 문자열을 탐색하여 반환한다.
print(result2)

```

```python
text = "apple banana cat"
result = re.findall(r"[at]", text)
if result: # result에 a 아니면 t가 있으면 반환되서 True가 됨
    print("a 아니면 t가 있다")
else:
    print("a와 t 둘다 없다")

# findall은 문자열을 찾아서 반환하므로 있다 없다만 알고 싶을 때 search method를 사용한다

result2 = re.search(r"[at]", text) # 있으면 match 객체 반환, 없으면 None 반환
if result2:
    print("a 아니면 t가 있다")
else:
    print("a와 t 둘다 없다")
```

```python
# 대괄호[] 안의 "-"는 범위를 나타낸다. 문자0부터 문자9까지 [0123456789]를 줄여서 [0-9]로 표현
# [a-z], [A-Z], ...
import re

text = "A12B34"
print(re.findall(r"[0-9]", text))
# ['1', '2', '3', '4']
# 문자0부터 문자9까지 하나라도 일치하면 반환되는 리스트에 추가 
```

```python
# 영문 대문자 찾기
text = "Python JAVA AI"
print(re.findall(r"[A-Z]", text))

# 영문 소문자 찾기
print(re.findall(r"[a-z]", text))

# 소문자 대문자 숫자 다 찾기
print(re.findall(r"[a-zA-Z0-9]", text))
```

```python
# 특정 문자 type 제외 [^] 주의할 것은 대괄호 밖의 ^은 시작을 나타냄
# 대괄호 안에 ^는 괄호안의 모든 문자를 제외한 문자
text = "a7bc123"
result = re.findall(r"[^0-9]", text) # 0~9문자가 아니면 반환되는 리스트에 추가
print(result)
```

```python
# search는 있으면 match, 객체 없으면 None 반환
# 비밀번호 규칙 : 소문자 대문자 포함, 특수문자 포함, 숫자문자 포함

import re

password = input("비밀번호 : ")

result = []

if re.search(r"[가-힣]", password):
    result.append("한글을 사용할 수 없습니다")

if not re.search(r"[a-z]", password):
    result.append("소문자가 없습니다.")

if not re.search(r"[A-Z]", password):
    result.append("대문자가 없습니다.")

if not re.search(r"[0-9]", password):
    result.append("숫자가 없습니다.")

if not re.search(r"[^a-zA-Z0-9가-힣]", password):
    result.append("특수문자가 없습니다.")

if len(result) == 0:
    result.append("사용 가능한 비밀번호입니다.")

for r in result:
    print(r)
```
