```python
# 시간복잡도

numbers = [15, 8, 23, 7, 31, 18]
target = 23

count = 0

for i in range(len(numbers)):
    count += 1
    if numbers[i] == target:
        print("찾은 위치 :", i)
        break # 찾으면 남은 원소는 비교하지않고 break

print("비교 횟수 :", count)        # 최대 n번 , O(n)
```

```python
numbers = [15, 8, 23, 7, 31, 18]

max_num = numbers[0]  # 또는 임의의 엄청나게 작은 값을 할당하여 첫번째 값이 최댓값이 되도록 하게 하는 두가지 방법이 있다.
count = 0

for n in numbers:
    count += 1
    if n > max_num:
        max_num = n

print("최댓값 :", max_num)
print("비교 횟수 :", count)    # 무조건 n번 , O(n)
```

```python
# bubble sort
numbers = [5, 3, 8, 1, 4]

# 데이터 5개
# 1회전 : 4번 비교   -> 맨 마지막(4 인덱스)에 제일 큰 숫자가 위치하게 됨
# 2회전 : 3번 비교  -> 3인덱스에 두번째로 큰 숫자가 위치하게 됨
# 3회전 : 2번 비교  -> 2인덱스에 세번째로 큰 숫자가 위치하게 됨
# 4회전 : 1번 비교  -> 1인덱스에 네번째로 큰 숫자가 위치하게 됨

count = 0

for j in range(len(numbers) - 1):
    for i in range(len(numbers) - 1 - j):
        count += 1
        if numbers[i] > numbers[i + 1]:
            numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i] # 우측 패킹 -> 대입 -> 좌측 언패킹
            # temp = numbers[i]
            # numbers[i] = numbers[i+1]
            # numbers[i+1] = temp

print(numbers)
print("비교 횟수 :", count)      # O(n^2)
 # 같은 O(n²) 알고리즘이라도 실제 비교 횟수와 실행 시간은 서로 다를 수 있다
# 이렇게 for 중첩문으로 짜는 것 보다 선형으로 코드 짜는게 좋다.
```

```python
# 실행 횟수 비교
import random
import time

# 0~9999 사이의 숫자 1000개를 무작위로 생성
numbers = random.sample(range(10000), 1000) # n을 2배로 늘려도 시간은 2배가 되진 않고 미미하다

# ===========================
# 선형 탐색 (O(n))
# ===========================

# 실행 시간 측정 시작
start = time.time() # start에 1970년 1월 1일 0시에 0.000초로 카운팅 시작, 이때로부터 현재 몇초가 지났는지 나옴

# 마지막 원소를 찾도록 설정(최악의 경우 : 최대 실행횟수)
target = numbers[-1]
count = 0

# 처음부터 끝까지 하나씩 비교
for n in numbers:
    count += 1
    if n == target:
        break

# 실행 시간 측정 종료
end = time.time()

print("선형 탐색 :", end - start, "count:", count)

# ===========================
# 버블 정렬 (O(n²))
# ===========================

# 새로운 무작위 데이터 생성
numbers = random.sample(range(10000), 1000)

# 실행 시간 측정 시작
start = time.time()

# 버블 정렬 수행                       # 메모리는 여전히 n이 1000이라 공간 복잡도는 변하지 않으나 시간복잡도가 훨씬 증가해서 훨씬 느려짐.
for j in range(len(numbers) - 1):      # n이 두배가 되면 시간이 대략 n^2배 증가한다.
    for i in range(len(numbers) - 1 - j):
        if numbers[i] > numbers[i + 1]:
            numbers[i], numbers[i + 1] = numbers[i + 1], numbers[i]

# 실행 시간 측정 종료
end = time.time()

print("버블 정렬 :", end - start)
```
