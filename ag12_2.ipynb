```python
# 공간복잡도 O(n)
n = int(input("숫자 개수 : "))

numbers = []

for i in range(n):
    numbers.append(i)

print(numbers)
```

```python
# 공간복잡도 O(n^2)
n = int(input("크기 : "))

matrix = []

for i in range(n): # for i in range(n) * range(n) 로 하면 12_1에서 했듯이 의미가 없다 공간복잡도는 여전히 O(n^2)
    row = []
    for j in range(n):
        row.append(0)
    matrix.append(row)

print(matrix)
```
