```python
# sliding window : 한칸 이동하면 그 원소를 더하고 이전 빠진 원소를 뺀 것이다. 즉 누적합 list를 이용하면 된다

scores = [10, 20, 3, 40, 5, 60, 8, 55, 44]

k = 3 # window size

# window 시작 포인트(인덱스 0)의 합을 초기값으로 설정
window = sum(scores[:k]) # index 0부터 2까지의 sum이 default 값 - > scores[0:3] = scores[0:k] = scores[:k]

totals = [window] # 초기값
avgs= [window/k] # 초기값
max_val = {"구간": f"0 ~ {k-1}" , "최대값":window} # 초기값
min_val = {"구간": f"0 ~ {k-1}" , "최소값":window} # 초기값

# 초기값에서 1칸을 이동 -> index k-1에서 k로 이동
# 첫번째 우측 이동 : window + scores[k] - score[k-k]
# 두번째 우측 이동 : window + scores[k+1] - score[k+1 - k]  이것을 아래의 반복문으로 구현하자

print(window) # 초기값 33

for i in range(k, len(scores)): # range(3, 6) , 3칸 이동 
    # 원도우크기 부터 1칸식 이동하며 윈도우안의 합을 연산
    window = window + scores[i]- scores[i-k] 
    # 처음 이동 : window + scores[3]- scores[3-3] -> 63
    # 두번째 이동 : window + scores[4] - scores[4-3] -> 48
    # 세번째 이동 : window + scores[5] - scores[5-3] -> 105


    totals.append(window)
    avgs.append(window/k)

    section = f"{i-(k-1)} ~ {i}"

    if window > max_val["최대값"]:
        max_val = {"구간": section , "최대값":window}
    if window < min_val["최소값"]:
        min_val = {"구간": section , "최소값":window}
    


print(totals)
print(avgs)
print(max_val)
print(min_val)
```
