```python
# tree 구조가 linear 구조보다 저장속도가 느리다, 양쪽 중에 어디로 갈지 정해야해서
# 찾는 속도는 tree 구조가 더 빠르다, linear는 최악의 경우 끝까지 다 비교해야 함.
# 즉 tree는 검색, 탐색에 좋은 자료구조
```

```python
nums = [ 1, 3, 5, 7, 9, 10, 20, 10]

# 첫번째부터 8번째까지 누적합 구하라
# index 0부터 7까지
t1 = 0
for i in range(len(nums)) : # index 0~7
    t1 += nums[i]


print(t1)



# 첫번째부터 세번째 까지

t2 = 0
for i in range(3): # 0~2
    t2 += nums[i]
print(t2)                   # 비효율적이다, 위에서 이미 전체를 구했는데 일부구하기 위해 다시 변수 만들고 하면 시간복잡도 및 공간복잡도 안좋아짐

# 구간별 누적합 list를 구하자

nums = [ 1, 3, 5, 7, 9, 10, 20, 10]

# 누적합 list는 default 0부터 시작하게 한다
prifix = [0]
for n in nums:
    prifix.append(prifix[-1]+n) # 누적합 list 마지막 index[-1]에다가 반복문의 현재 n 원소를 더해주면 누적합

```

```python
prifix[8]  # 누적합 list는 처음에 0 default이므로 , 사실 prifix[8] - prifix[0]임
s = 5
e = 7
# 원본의 5번째~ 7번째까지의 누적합 구해보자
prifix[e]-prifix[s-1]

# 위와같이 누적합 list를 구해놓으면 반복문을 처음에 생성할 때 한번만 사용하면 되고 메모리 공간은 n+1만 더 생성하면 된다.
# 이렇게 안 구해놓으면 시간복잡도가 엄청 복잡해질 수 있다.
```
