```python
stack = [] # python 리스트에서 stack 구조의 push/pop과 동일한 method가 있다.

stack.append(10)
stack.append(20)
stack.append(30)

print(stack)

num = stack.pop()  # 30이 빠져서 num에 저장됨

print(num)
print(stack)

num = stack.pop()  # 20이 빠져서 num에 저장됨

print(num)
print(stack)

num = stack.pop() # 10이 빠져서 num에 저장됨

print(num)
print(stack)
```

```python
# 웹 브라우저 "뒤로가기" 구현
stack = []

stack.append("네이버")
stack.append("구글")
stack.append("ChatGPT")

print("현재 방문 기록 :", stack)
print("뒤로가기 :", stack.pop())
print("현재 페이지 :", stack[-1])
```

```python
plates = []
plates.append({"color" : "red", "price" : 100 })
plates.append({"color" : "blue", "price" : 150})
plates.append({"color" : "blue", "price" : 150})
plates.append({"color" : "black", "price" : 200})
plates.append({"color" : "red", "price" : 100})

print("plates")

total_price = 0

while (len(plates)) > 0 :
    p = plates.pop()
    total_price += p["price"]

print(f"{total_price : ,}")
print(plates)
```
