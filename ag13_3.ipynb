```python
# 조건 분기 프로세스를 하나의 트리 자료구조형태로 구현한 코드
# yes 안에 quesition, yes, no 그리고 no 안에 question, yes, no가 있음을 알 수 있다

# stack은 list, tree는 dictionary

fruit_tree = {
    "question": ("color", "Red"), # 보통 표현식은 tuple로 되있다

    "yes": {
        "question": ("size", "Large"),
        "yes": "사과 (Apple)",
        "no": "체리 (Cherry)"
    },

    "no": {
        "question": ("color", "Purple"),
        "yes": "포도 (Grape)",
        "no": "모르는 과일"
    }
}
# 더이상 dictionary 형태가 나오지 않을 때까지 반복하게 됨
# 더 조건이나 원소가 추가되면 dictionary를 추가하면 된다
# 코드 상에서는 if, else 등이 안나타낮니만 숨겨져있다.
# tree는 병렬로 읽을 수 있다 (?) 짤라서
```

```python
# 트리 탐색 코드
# isinstance(a,b)->bool 함수는 변수 a가 b타입의 객체(instance)이면 True 아니면 False를 반환하는 함수

fruit = {
    "color": "Red",
    "size": "Large" # Small
}

fruit_tree = {
    "question": ("color", "Red"),

    "yes": {
        "question": ("size", "Large"),
        "yes": "사과 (Apple)",
        "no": "체리 (Cherry)"
    },

    "no": {
        "question": ("color", "Purple"),
        "yes": "포도 (Grape)",
        "no": "모르는 과일"
    }
}
node = fruit_tree # node 변수의 초기값은 dictionary -> 무한루프 -> node 변수값이 dictionary가 아닐 때까지

# 트리를 탐색하는 반복문
while isinstance(node, dict): # node객체가 dict의 instance 이라면 while 아래 블록 실행되고 dict이 아니면 while문을 빠져나오게 됨

    # dict이라면 "question"키값을 찾아 질문은 key, 정답은 value에 저장 
    key, value = node["question"] 
   
    # 입력데이터의 key값과 비교하여 정답이면 "yes"키값을 node에 대입, 아니면 "no"키값을 node에 대입
    # 대입된 node값이 dict 타입이면 다시 반복문 진행... 
    # node에 dict타입이 아닌 타입(위 트리에선 문자열)이 올때 까지 반복문은 계속
    if fruit[key] == value: # 맨 위 입력값 fruit의 key true, false값 반환위한 코드
        node = node["yes"]
    else:
        node = node["no"]

# 반복문을 빠져 나오면 마지막 node값을 출력
print(node)
```
