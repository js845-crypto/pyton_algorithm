```python
# 문제 3: 상품별 판매 수 계산

count = {} 
# 노트북이 없으면 count dictionary에 추가, 있으면 노트북 항목에 +1
for r in sales:
    key_name = r["상품"]
    if key_name not in count :
        count[key_name] = 0 # count에 key_name이 없으면 추가
    
    count[key_name] += 1 # key_name이 count에 있던 없던 +1 수량 추가해야하므로 코드 하나 더 줄이기 위해 if 문 블록 안에 안 넣음
print(count)
for k , v in count.items():
    print(f"{k} : {v}건 입니다")
```

```python
# 문제 8 : 이동평균 
#  판매액    sliding window 
# sliding window에 사용될 list 뽑아내자

prices = []
for r in sales:
    if isinstance(r["판매액"], int):
        prices.append(r["판매액"])

prices

window = 3

# [0:3] -> [1:4] -> [2:5]
for i in range(len(prices) - (window - 1)) :
    total = sum(prices[i: i+ window])
    avg = round(total / window , 3 )
    print(f"{i+1} ~ {i+window} 구간의 평균 : {avg}원")
```

```python
# 문제 9 : 누적 평균 , 1일 평균 -> 1~2일 평균 -> 1~3일 평균 - > 1~4일 평균 . . . 
# 먼저 누적 합 list를 구한 후에 index로 나눠주면 누적 평균들을 원소로 갖는 list가 나옴

prices = []
for r in sales:
    if isinstance(r["판매액"], int):
        prices.append(r["판매액"])

# 위의 코드는 먼저 판매액 숫자인것들만 추출

prefix = [0]

for p in prices:
    prefix.append(prefix[-1] + p)

avgs = []

for i in range(1,len(prefix)):
    avgs.append(prefix[i]/i)

avgs







```

```python
# 문제 11 : 결측치 처리 , "", "결측"을 중앙값으로 처리하자
# 평균 sum(prices) / len(prices)

# print(len(prices)) 짝수개이므로 중앙 두개의 값을 2로 나눈 것이 중앙값이 됨 곤란해짐

# 짝수개 홀수 개 케이스 두개로 나누어야함 분기!

prices.sort() # prices list를 메소드 sort사용해서 정렬해서 원본 수정
print(prices)
median = 0
if len(prices) % 2 ==1 :
    median = prices[len(prices)//2] # 중앙값 1개
else:
    median = (prices[(len(prices)-1)//2] + prices[len(prices)//2])/2

print(f"중앙값 : {median}")
print("원본" , sales)

for r in sales:
    if not isinstance(r["판매액"], int):
        r["판매액"] = median

print("결측치 -> 중앙값" , sales)

```

```python
# 문제 12 : 범주별 통계
result = {}
for r in sales:
    p_name = r["상품"]
    price = r["판매액"]
    total_count = r["수량"]
    if r["상품"] not in result:
        # result[r["상품"]] = {"합계" : 0,} r["상품"]이 헷갈리므로 p_name으로 대체하자 변수도입해서
        
        result[p_name] = {"합계" : 0, "건수" : 0, "최대": price, "최소" : price, "총 수량" : 0} # 최대 최소값은 default로 처음 값을 설정.

        # "키보드" : {"합계" : 0, "건수" : 0, "최대":0, "최소" : 0, "총 수량" : 0}
        # "모니터" : {"합계" : 0, "건수" : 0, "최대":0, "최소" : 0, "총 수량" : 0}

    info = result[p_name]

    info["합계"] += price # r["판매액"]이 불편해서 price 위의 변수 도입해서 치환 , info도 마찬가지로 치환위해서 도입
    info["건수"] += 1
    info["총 수량"] += total_count

    # 최대/ 최소

    if info["최대"] < price :
        info["최대"] = price

    if info["최소"] > price:
        info["최소"] = price 

print(result)

```
